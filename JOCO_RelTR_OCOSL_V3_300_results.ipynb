# JOCO — OI-trained RelTR vs hierarchy-aware OCOSL V3

Final 300-image controlled experiment. RelTR-only and RelTR+OCOSL use the **same learned candidate pool**. The V3 change is limited to hierarchy-aware domain/range compatibility using transitive `rdfs:subClassOf` closure.

**Final headline:** RelTR-only F1 = **0.459130**, OCOSL V3 F1 = **0.458503**, ΔF1 = **-0.000628**, paired permutation p = **1.0**, and 298/300 images are unchanged.


````python
from pathlib import Path
import json, pandas as pd, numpy as np

# Works from the GitHub package root or a Kaggle working directory.
ROOT = Path.cwd()
RESULTS = ROOT / 'results'
if not RESULTS.exists():
    RESULTS = ROOT

summary_path = RESULTS / 'oi_reltr_ocosl_controlled_300_v3_hierarchy.json'
csv_path = RESULTS / 'oi_reltr_ocosl_controlled_300_v3_hierarchy.csv'
relation_path = RESULTS / 'reltr_ocosl_relation_impact_300_v3.csv'
image_path = RESULTS / 'reltr_ocosl_image_impact_300_v3.csv'

summary = json.loads(summary_path.read_text(encoding='utf-8'))
df = pd.read_csv(csv_path)
relation_impact = pd.read_csv(relation_path)
image_impact = pd.read_csv(image_path)
print('Rows:', len(df))
````

````python
table = pd.DataFrame([
    {'Method': 'Raw OI-RelTR', **summary['raw_oi_reltr']},
    {'Method': 'RelTR-only, shared pool', **summary['reltr_only_shared_pool']},
    {'Method': 'RelTR + OCOSL V3', **summary['reltr_plus_ocosl_v3']},
])
table[['Method','TP','FP','FN','Precision','Recall','F1']]
````

````python
print('micro F1 gain:', summary['micro_f1_gain'])
print('paired permutation p:', summary['paired_permutation_pvalue'])
print('improved/equal/worse:', summary['per_image_improved'], summary['per_image_equal'], summary['per_image_worse'])
print('fallback count:', summary['fallback_count'])
print('avg solver time (s/image):', summary['avg_solver_time_s'])
print('canonicalized endpoints:', summary['canonicalized_endpoint_count'])
````
